```python
arr1 = np.random.randint(1, 101, 10)
res1 = pd.Series(arr1)
print(res1)
```

```python
print(res1.max())
print(res1.min())
print(res1.mean())
```

```python
res2 = res1 ** 2
print(res2)
```
