```python
import numpy as np
import pandas as pd
```

```python
data = {
    
    "name" : ["alice", "bob","alice", "charlie", "bob", "david"],
    "age" : [24,27,24,22,27,32],
    "city" : ["New York", "Los Angeles","New York", "Chicago","Los Angeles", "Houston"]
}

df = pd.DataFrame(data)
print("handling duplicate data")
print("original data")
print(df)
```

```python
print(df.duplicated())
```

```python
print(df[df.duplicated()])
```

```python
print("drop duplicates keep first")
df_first = df.drop_duplicates()
print(df_first)
```

```python
df_last = df.drop_duplicates(keep='last')
print(df_last)
```

```python
data = {
    
    "name" : ["alice", "bob","alice", "charlie", "bob", "david"],
    "age" : [24,np.nan,24,22,np.nan ,32],
    "city" : ["New York", "Los Angeles",np.nan , "Chicago","Los Angeles", np.nan],
    'salary' : [50000,60000,np.nan, np.nan,60000,70000]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
print("detect NaN values")
print(df.isna())

print("count no. of NaN values per column")
print(df.isna().sum())
```

```python
print("drop rows with NaN values")

df_drop_rows = df.dropna()

print(df_drop_rows)
```

```python
print(df.dropna(how='all'))
```

```python
print("drop NaN in Age and Salary")
print(df.dropna(subset=['age', 'salary']))

```

```python
print("Fill NaN values")
df_fill = df.copy()

df_fill['age'] = df_fill['age'].fillna(0)
df_fill['city'] = df_fill['city'].fillna('unknown')
df_fill['salary'] = df_fill['salary'].fillna(0)

print(df_fill)
```

```python
print("fill col. with mean value")


df_fillmean = df.copy()
df_fillmean['age'] = df_fillmean['age'].fillna(df_fillmean['age'].mean())
df_fillmean['salary'] = df_fillmean['salary'].fillna(df_fillmean['salary'].mean())
print(df_fillmean)
```

```python
print("forward fill")
df_interpolate=df.copy()
df_interpolate['age']=df_interpolate['age'].ffill()
df_interpolate['salary']=df_interpolate['salary'].ffill()
print(df_interpolate)
```
