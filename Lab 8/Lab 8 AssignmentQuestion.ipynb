create df using dict,a) select rows age>25,b) select rows salary b/w x and y

```python
import numpy as np
import  pandas as pd
```

```python
data ={
    'Name':['Amit','Riya','John','Sara'],
    'Age':[25,30,22,28],
    'Salary':[50000,60000,55000,65000]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
print(df[df['Age']>25])
```

```python
print(df[df['Salary'].between(55000,65000)])
```
